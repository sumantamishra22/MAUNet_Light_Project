```python
MAUNet_Test_Mean=np.load(path+'/Result/SpatialMean/Work7_SpatialMean_MAUNet_TRMMIMD_TestResult.npy')
MAUNet_Light_KR_Test_Mean=np.load(path+'/Result/SpatialMean/Work7_SpatialMean_MAUNet_Light_KR_TRMMIMD_TestResult.npy')
Label_Test_Mean=np.load(path+'/Result/SpatialMean/Work7_SpatialMean_Label_Test_TRMMIMD.npy')
```

```python
# List of data and corresponding labels for each method
data_list = [
    (Label_Test_Mean, 'Observation'),
    (MAUNet_Test_Mean, 'MAUNet'),
    (MAUNet_Light_KR_Test_Mean, 'MAUNet_Light_KR'),
]

# Create subplots: one row for each method
fig, axes = plt.subplots(nrows=len(data_list), ncols=1, figsize=(5, 7))

# Set the same x-axis range and y-axis range for all plots
x_min, x_max = 0, 25
y_min, y_max = 0, 20

# Loop through the data and create individual histogram plots
for i, (data, label) in enumerate(data_list):
    axes[i].hist(data, bins=200, range=(x_min, x_max), color='blue', edgecolor='black', alpha=0.7)
    #axes[i].set_xlim([x_min, x_max])
    axes[i].set_ylim([y_min, y_max])
    axes[i].set_xlabel('Spatial Mean Rainfall')
    axes[i].set_ylabel('Frequency')
    axes[i].set_title(label)
    axes[i].legend([label])

# Adjust layout for better spacing
plt.tight_layout()

# Show the plot
plt.show()
```
